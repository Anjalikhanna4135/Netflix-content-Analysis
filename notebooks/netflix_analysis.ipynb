```python
import pandas as pd

df = pd.read_csv("../data/raw/netflix_titles.csv")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['listed_in'] = df['listed_in'].apply(lambda x: x.split(', '))
```

```python
df_exploded = df.explode('listed_in')
```

```python
df_exploded = df_exploded.reset_index(drop=True)
```

```python
df_exploded.head()
```

```python
df['country'] = df['country'].apply(
    lambda x: x.split(', ') if isinstance(x, str) else ['Unknown']
)
```

```python
df_country = df.explode('country').reset_index(drop=True)
```

```python
df['country'].head(20)
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
```

```python
df[['date_added', 'year_added']].head()
```

```python
df['year_added'].value_counts().sort_index()
```

```python
df.to_csv("../data/processed/netflix_cleaned.csv", index=False)
```

```python
df_exploded.to_csv("../data/processed/netflix_genres_exploded.csv", index=False)
```

```python
df_country.to_csv("../data/processed/netflix_countries_exploded.csv", index=False)
```
